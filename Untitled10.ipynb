```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sn
```

**Q1. Handle Missing Values (Basic Cleaning)**

Fill missing values using the following:

* Age → Median

* Salary → Mean

* Hours_Worked_Per_Week → Median

* Performance_Score → Mean

Display the dataset after handling missing values.

```python
# Check missing values before handling
print("Missing values before handling:")
print(df[[
    "Age",
    "Salary",
    "Hours_Worked_Per_Week",
    "Performance_Score"
]].isnull().sum())

# Fill missing values
df["Age"] = df["Age"].fillna(df["Age"].median())

df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

df["Hours_Worked_Per_Week"] = df["Hours_Worked_Per_Week"].fillna(
    df["Hours_Worked_Per_Week"].median()
)

df["Performance_Score"] = df["Performance_Score"].fillna(
    df["Performance_Score"].mean()
)

# Display missing values after handling
print("\nMissing values after handling:")
print(df[[
    "Age",
    "Salary",
    "Hours_Worked_Per_Week",
    "Performance_Score"
]].isnull().sum())

# Display dataset
display(df)
```

**Q2. Label Encoding**

Convert these columns into numeric values using Label Encoding:

* Gender
* Department

Show the updated columns.

```python

from sklearn.preprocessing import LabelEncoder

# Create LabelEncoder
le_gender = LabelEncoder()
le_department = LabelEncoder()

# Apply encoding
df["Gender"] = le_gender.fit_transform(df["Gender"])

df["Department"] = le_department.fit_transform(df["Department"])

# Display updated columns
print("Updated Gender and Department columns:")
display(df[["Gender", "Department"]])

# Display mappings
print("\nGender Mapping:")
print(dict(zip(
    le_gender.classes_,
    le_gender.transform(le_gender.classes_)
)))

print("\nDepartment Mapping:")
print(dict(zip(
    le_department.classes_,
    le_department.transform(le_department.classes_)
)))
```

**Q3. One-Hot Encoding**

Apply One-Hot Encoding on:

* Work_Mode

* Location

Display the dataset and check how many new columns are created.

```python
from sklearn.preprocessing import OneHotEncoder

# Create encoder
encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore"
)

# Apply One-Hot Encoding
encoded_data = encoder.fit_transform(
    df[["Work_Mode", "Location"]]
)

# Get new column names
new_columns = encoder.get_feature_names_out(
    ["Work_Mode", "Location"]
)

# Convert encoded data into DataFrame
encoded_df = pd.DataFrame(
    encoded_data,
    columns=new_columns,
    index=df.index
)

# Remove original categorical columns
df_onehot = df.drop(
    columns=["Work_Mode", "Location"]
)

# Combine original dataset with encoded columns
df_onehot = pd.concat(
    [df_onehot, encoded_df],
    axis=1
)

# Display dataset
display(df_onehot)

# Number of new columns
print("Number of new columns created:", len(new_columns))

print("\nNew columns:")
print(list(new_columns))
```

Normalize the following columns:

Use MinMaxScaler and display results.

Q7. Build Preprocessing Pipeline

Create a pipeline that:

* Applies encoding to categorical columns

* Applies scaling to numerical columns

Use:

* ColumnTransformer

* Pipeline

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Define categorical columns
categorical_columns = [
    "Gender",
    "Department",
    "Work_Mode",
    "Location"
]

# Define numerical columns
numerical_columns = [
    "Age",
    "Salary",
    "Hours_Worked_Per_Week",
    "Performance_Score",
    "Projects_Completed"
]

# Categorical pipeline
categorical_pipeline = Pipeline(
    steps=[
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

# Numerical pipeline
numerical_pipeline = Pipeline(
    steps=[
        (
            "scaler",
            StandardScaler()
        )
    ]
)

# ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            categorical_pipeline,
            categorical_columns
        ),
        (
            "numerical",
            numerical_pipeline,
            numerical_columns
        )
    ]
)

# Complete Pipeline
pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        )
    ]
)

print("Preprocessing Pipeline created successfully.")

display(pipeline)
```

**Q8. Apply Pipeline**

Apply the pipeline on the dataset.

Display:

* Transformed dataset

* Shape of final dataset

```python
# Apply the pipeline
transformed_data = pipeline.fit_transform(df)

# Get transformed column names
feature_names = pipeline[
    "preprocessor"
].get_feature_names_out()

# Convert transformed data into DataFrame
transformed_df = pd.DataFrame(
    transformed_data,
    columns=feature_names,
    index=df.index
)

# Display transformed dataset
print("Transformed Dataset:")
display(transformed_df)

# Display shape
print("\nShape of final dataset:")
print(transformed_df.shape)
```

**Q9. Conceptual Question**

Why is scaling important in Python?

Solution:

Scaling is important because different numerical features may have different ranges. A feature with very large values can have more influence on some machine-learning algorithms. Scaling brings numerical features to a comparable range and can improve model performance and stability.

For example, Salary may contain values in thousands while Performance Score may contain values from 1 to 10.

**Q10. Conceptual Question**

Why do we convert categorical data into numerical form?

Solution:

Machine-learning algorithms generally work with numerical data. Categorical values such as Gender, Department, Work Mode and Location are therefore converted into numerical representations. Label Encoding assigns numerical labels to categories, while One-Hot Encoding creates separate binary columns for each category.
